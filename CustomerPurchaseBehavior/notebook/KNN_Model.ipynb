### ⚙️ Function / Method Arguments Explanation
**Function:** `train_test_split`
**Argument 1:** `test_size=0.3`
- **3.1 What it does:** Holds back 30% of data.
- **3.2 Why it is used:** We need enough questions for the final exam.
- **3.3 When to use it:** Always.
- **3.4 Where to use it:** Function call.
- **3.5 How to use it:** `0.3`.
- **3.6 Internal Effect:** Slicing index.
- **3.7 Output impact:** Size of X_test.

**Argument 2:** `stratify=y`
- **3.1 What it does:** Keeps class ratios same.
- **3.2 Why it is used:** To prevent "Missing Class" errors.
- **3.3 When to use it:** Classification.
- **3.4 Where to use it:** Function call.
- **3.5 How to use it:** `stratify=y`.
- **3.6 Internal Effect:** Grouped shuffle.
- **3.7 Output impact:** Fair distribution.

```python
X = df.drop('PurchaseCategory', axis=1)
y = df['PurchaseCategory']

numeric_features = ['Age', 'Income', 'MonthlySpending', 'SessionDuration', 'PageViewsPerVisit', 'AccountAge']
categorical_features = ['DeviceType', 'MembershipTier']

preprocessor = ColumnTransformer(
    transformers=[
        ('num', Pipeline([('imputer', SimpleImputer(strategy='mean')), ('scaler', StandardScaler())]), numeric_features),
        ('cat', Pipeline([('onehot', OneHotEncoder(handle_unknown='ignore'))]), categorical_features)
    ])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
```

# PART B: MODEL IMPLEMENTATION

### 🔹 Line Explanation
#### 2.1 What the line does
Builds the KNN Pipeline.
#### 2.2 Why it is used
To automate cleaning + prediction.
#### 2.3 When to use it

#### 2.5 How to use it

### ⚙️ Function / Method Arguments Explanation
**Function:** `KNeighborsClassifier`
**Argument 1:** `n_neighbors=5`
- **3.1 What it does:** Sets 'k'.
- **3.2 Why it is used:** To define how many neighbors vote.
- **3.3 When to use it:** Tuning the model.
- **3.4 Where to use it:** Constructor.
- **3.5 How to use it:** `int`.
- **3.6 Internal Effect:** Algorithm constraint.
- **3.7 Output impact:** Smoothness of decision boundary.

```python
knn_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', KNeighborsClassifier(n_neighbors=5))
])

print("Training...")
knn_pipeline.fit(X_train, y_train)
print("Done!")
```

```python
y_pred = knn_pipeline.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Accuracy: {acc:.4f}")
```
