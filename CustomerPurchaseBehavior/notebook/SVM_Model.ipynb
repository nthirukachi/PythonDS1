```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.impute import SimpleImputer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.svm import SVC
from sklearn.metrics import classification_report, confusion_matrix, accuracy_score
```

```python
FILE_PATH = 'C:/nagpython/demouv/CustomerPurchaseBehavior/data/customer_behavior.csv'
df = pd.read_csv(FILE_PATH)
X = df.drop('PurchaseCategory', axis=1)
y = df['PurchaseCategory']
```

### 🔹 Line Explanation (Feature Scaling)
#### 2.1 What the line does
Defines `StandardScaler` in pipeline.
#### 2.2 Why it is used
**CRITICAL:** SVM relies on geometry. Unscaled features (Income vs Age) distort the space.
#### 2.3 When to use it
Use it or SVM will fail.
#### 2.4 Where to use it
Pipeline.
#### 2.5 How to use it
`StandardScaler()`
#### 2.6 How it works internally
Subtract Mean, Divide by Std Dev.
#### 2.7 Output with sample examples
Similar range features.

```python
numeric_features = ['Age', 'Income', 'MonthlySpending', 'SessionDuration', 'PageViewsPerVisit', 'AccountAge']
categorical_features = ['DeviceType', 'MembershipTier']

numeric_transformer = Pipeline(steps=[
    ('imputer', SimpleImputer(strategy='mean')),
    ('scaler', StandardScaler()) # Mandatory for SVM
])

categorical_transformer = Pipeline(steps=[
    ('onehot', OneHotEncoder(handle_unknown='ignore'))
])

preprocessor = ColumnTransformer(
    transformers=[
        ('num', numeric_transformer, numeric_features),
        ('cat', categorical_transformer, categorical_features)
    ])

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, stratify=y, random_state=42)
```

# PART B: MODEL IMPLEMENTATION

### ⚙️ Function / Method Arguments Explanation
**Function:** `SVC`
**Argument 1:** `kernel='rbf'`
- **3.1 What it does:** Selects the math function to measure distance.
- **3.2 Why it is used:** To handle curved interactions between variables.
- **3.3 When to use it:** When data isn't a straight line.
- **3.4 Where to use it:** Constructor.
- **3.5 How to use it:** `kernel='linear'` or `kernel='rbf'`.
- **3.6 Internal Effect:** Projects data to Hilbert Space.
- **3.7 Output impact:** Ability to classify complex patterns.

```python
svm_pipeline = Pipeline(steps=[
    ('preprocessor', preprocessor),
    ('classifier', SVC(kernel='rbf', C=1.0, gamma='scale', random_state=42))
])

print("Training SVM...")
svm_pipeline.fit(X_train, y_train)
print("Done!")
```

```python
y_pred = svm_pipeline.predict(X_test)
acc = accuracy_score(y_test, y_pred)
print(f"Correctness: {acc:.4f}")
```
